Ejercicio 1 — Ventas totales de marzo

Filtra el dataset para quedarte solo con marzo y calcula las ventas totales del mes. Compara el resultado con el de febrero, que se vio en clase ($424.650). Antes de calcular, revisa si hay algo en los datos que debería excluirse de una venta "neta" (devoluciones, cancelaciones) — si no hay ninguna columna que indique eso, acláralo como supuesto en tu respuesta.

```python
import pandas as pd

# Cargar el archivo 
df = pd.read_csv('../ventas_kiosco.csv')

# 1. Asegurarnos de que la columna 'fecha' (en minúscula) sea interpretada como fecha
df['fecha'] = pd.to_datetime(df['fecha'])

# 2. Filtrar el DataFrame para quedarnos solo con el mes de marzo (mes 3)
df_marzo = df[df['fecha'].dt.month == 3].copy()

# 3. Crear una nueva columna con el total multiplicando 'cantidad' y 'precio_unitario'
df_marzo['Total_Venta'] = df_marzo['cantidad'] * df_marzo['precio_unitario']

# 4. Sumar toda la columna para obtener el total mensual
ventas_totales_marzo = df_marzo['Total_Venta'].sum()

# 5. Mostrar el resultado para compararlo con febrero ($424.650)
print(f"Las ventas totales netas de marzo fueron: ${ventas_totales_marzo:,.2f}")
```

Ejercicio 2 — Ticket promedio de marzo
Con el resultado del Ejercicio 1 y la cantidad de compras de marzo, calcula el ticket promedio. Compáralo con el de febrero (~$2.831). Calcula también la mediana de las ventas de marzo y compara ambos valores — si son muy distintos, explica qué podría estar causando esa diferencia.

```python
# 1. Calcular el ticket promedio (la media)
ticket_promedio_marzo = df_marzo['Total_Venta'].mean()

# 2. Calcular la mediana de las ventas
mediana_marzo = df_marzo['Total_Venta'].median()

# 3. Mostrar los resultados para compararlos
print(f"Ticket Promedio de marzo (Media): ${ticket_promedio_marzo:,.2f}")
print(f"Mediana de ventas de marzo: ${mediana_marzo:,.2f}")
```

Ejercicio 4 — Tasa de churn de febrero a marzo
De los clientes que compraron en febrero, cuenta cuántos no volvieron a comprar en marzo. Calcula la tasa de churn. Antes de calcularla, escribe en una oración qué ventana de tiempo estás asumiendo como "cliente perdido" y por qué te parece razonable para este negocio.

Identificar  a los clientes unicos de Febrero


```python
import pandas as pd
import tkinter as tk
from tkinter import filedialog

# 1. Abrir ventana de Windows para buscar el archivo manualmente
root = tk.Tk()
root.attributes('-topmost', True)
root.withdraw() 
ruta_exacta = filedialog.askopenfilename(title="Busca y selecciona tu archivo ventas_kiosco")

print(f"La ruta real de tu archivo es: {ruta_exacta}")

# 2. Cargar el archivo (Volvemos a read_csv)
try:
    # Intentamos leerlo como un CSV normal
    df = pd.read_csv(ruta_exacta) 
except Exception:
    # Si falla, intentamos leerlo con separador de punto y coma
    df = pd.read_csv(ruta_exacta, sep=';')

df['fecha'] = pd.to_datetime(df['fecha'])

# 3. Cálculo de Churn
clientes_febrero = df[df['fecha'].dt.month == 2]['cliente'].unique()
clientes_marzo = df[df['fecha'].dt.month == 3]['cliente'].unique()

clientes_perdidos = set(clientes_febrero) - set(clientes_marzo)

total_febrero = len(clientes_febrero)
total_perdidos = len(clientes_perdidos)

if total_febrero > 0:
    tasa_churn = (total_perdidos / total_febrero) * 100
else:
    tasa_churn = 0

print("-" * 30)
print(f"Total clientes únicos en febrero: {total_febrero}")
print(f"Clientes que no volvieron en marzo (Perdidos): {total_perdidos}")
print(f"Tasa de Churn: {tasa_churn:.2f}%")
```
